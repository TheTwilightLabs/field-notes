## Completed solution

```python
import pandas as pd
import matplotlib.pyplot as plt
sales = pd.read_csv(DATASET_PATH)
by_region = sales.groupby('region')['revenue'].sum()
fig, ax = plt.subplots()
by_region.plot.bar(ax=ax, color='#164bff')
ax.set(title='Revenue by region', xlabel='Region', ylabel='Revenue')
plt.close(fig)
print('chart created')
```
