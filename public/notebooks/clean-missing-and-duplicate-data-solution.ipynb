## Completed solution

The cells below contain one complete, executable solution.


```python
import pandas as pd
learners = pd.read_csv(DATASET_PATH)
clean = learners.drop_duplicates().copy()
clean['study_hours'] = clean['study_hours'].fillna(clean['study_hours'].median())
assert clean['study_hours'].notna().all()
print(clean.shape)
```
