## Completed solution

```python
import pandas as pd
import matplotlib.pyplot as plt
learners = pd.read_csv(DATASET_PATH)
fig, ax = plt.subplots()
ax.scatter(learners['study_hours'], learners['final_score'], color='#164bff')
ax.set(title='Study time and final score', xlabel='Study hours', ylabel='Final score')
plt.close(fig)
print('chart created')
```
